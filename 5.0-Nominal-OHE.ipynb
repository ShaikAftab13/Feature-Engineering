### Nominal/OHE Encoding
One hot encoding, also known as nominal encoding, is a technique used to represent categorical data as numerical data, which is more suitable for machine learning algorithms. In this technique, each category is represented as a binary vector where each bit corresponds to a unique category. For example, if we have a categorical variable "color" with three possible values (red, green, blue), we can represent it using one hot encoding as follows:

1. Red: [1, 0, 0]
2. Green: [0, 1, 0]
3. Blue: [0, 0, 1]

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.DataFrame({
    'color': ['red','blue','red','green','green','blue','blue','green']
})
```

```python
df.head()
```

```python
encoder = OneHotEncoder()
```

```python
encoded = encoder.fit_transform(df[['color']]).toarray()
```

```python
encoded_df = pd.DataFrame(encoded,columns=encoder.get_feature_names_out())
```

```python
encoded_df.astype(int)
```

```python
encoder.transform([['blue']])
```

```python
df_encoded = pd.get_dummies(df['color'])
df_encoded.astype(int)
```

```python
pd.concat([df,encoded_df.astype(int)],axis=1)
```

```python
import seaborn as sns

df = sns.load_dataset('tips')
```

```python
df.head()
```

```python
df['sex'].value_counts()
```

```python
encoder = OneHotEncoder()

encoded = encoder.fit_transform(df[['sex','smoker','day','time']]).toarray()
```

```python
encoded_df = pd.DataFrame(encoded, columns = encoder.get_feature_names_out()).astype(int)
encoded_df.head()
```

```python
final_df = pd.concat([df,encoded_df],axis=1)
```
